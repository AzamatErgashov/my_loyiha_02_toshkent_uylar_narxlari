```python
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/main/housing_data_08-02-2021.csv')
data.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder 

ordinal = OrdinalEncoder()
joylashuv_category_encoded = ordinal.fit_transform(data[['district']])
joylashuv_category_encoded[:10]
```

```python
data['joylashuv_to_number'] = pd.DataFrame(joylashuv_category_encoded)
```

```python
column = ['rooms',	'size',	'level',	'max_levels',	'price',	'joylashuv_to_number']
data_number = data[column]
```

```python
from sklearn.base import BaseEstimator, TransformerMixin

class CombinedAttributesAdder(BaseEstimator, TransformerMixin):
    def __init__(self, clean_data):
        self.clean_data = clean_data
    def fit(self,df, y=None):
        return self
    def transform(self, df):
        df['price'] = df['price'].str.replace('Договорная', '0')
        df['price'] = df['price'].fillna(method='bfill')
        df['price'] = df['price'].astype(int)
        df['size'] = df['size'].str.replace('Площадьземли:1сот', '0').astype(float)
        return df
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler #yoki min_max_scalerdan ham foydalansa buladi

num_pipeline = Pipeline([# 1-qadam
    ('AttributesAdder', CombinedAttributesAdder(data)),
    ('min_max_bir_deopazonga_tushurish', MinMaxScaler())
])
num_pipeline.fit_transform(data_number)
num_pipeline
```

```python
num_pipeline
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder 
num_atributs = list(data_number)
str_atributs = ['district']

full_pipeline = ColumnTransformer([
    ('num', num_pipeline, num_atributs),# 1-nomi 2-pipelineni choqiramiz 3-qaysi ustunlar ustida pipeline ishlaganini aytypapmiz
    ('categorical', OrdinalEncoder(), str_atributs)
])
```

```python
data_prepared = full_pipeline.fit_transform(data)
```
