```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
```

```python
data = pd.read_csv('clean_data.csv', index_col=0)
```

```python
data.head()
```

```python
fig = px.bar(data, x="rooms", y="level_per_size", color="district_num", title="Long-Form Input")
fig.show()
```

```python
fig = px.scatter(data, x="district_num", y="price", color="price", title="Long-Form Input")
fig.show()
```

```python
plt.figure(figsize=(16,6))
sns.barplot(x=data['district_num'], y=data['price'])
plt.xticks(rotation=90)
plt.show()
```

```python
fig, ax = plt.subplots(2,2, figsize=(16,10))
fig.suptitle('grafiklar')
sns.histplot(ax=ax[0,0], data=data, x='rooms')
sns.histplot(ax=ax[0,1], data=data, x='district_num')
sns.scatterplot(ax=ax[1,0], data=data, x='district_num',y='price', hue='rooms', s=50)
sns.scatterplot(ax=ax[1,1], data=data, x='district_num', y='size',hue='rooms', s=50)
ax[0,0].set_title('price')
ax[0,1].set_title('year')
ax[1,0].set_title('year + price')
ax[1,1].set_title('tax + year')

plt.show()
```

```python
plt.figure(figsize=(7, 6))
plt.title('The Beer Regions', fontsize = 10)
sns.countplot(y='district_num', data=data)
```

```python
plt.figure(figsize = (18, 8))
plt.title('Dispensation of BeerScore', fontsize=20)
sns.histplot(data=data, x="district_num", kde=True, color= "purple")
plt.show()
```

```python
import plotly.express as px

fig = px.scatter_3d(data, x='district_num', y='price', z='rooms',
              color='district_num')
fig.show()
```

```python
def compute_correlations_matrix(dataset):
    corrMatrix = dataset.corr()
    plt.figure(figsize=(20,15))
    sns.heatmap(corrMatrix, annot=True, cmap='vlag')
    plt.show()

compute_correlations_matrix(data)
```
