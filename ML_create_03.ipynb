```python
import pandas as pd 
import sklearn
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
from IPython.display import Image 
from sklearn.model_selection import train_test_split    
%matplotlib inline
```

```python
import pandas as pd
data = pd.read_csv('housing_data_08-02-2021.csv', index_col=0)
data.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder 

category = ['district']
ordinal = OrdinalEncoder()
housing_category_encoded = ordinal.fit_transform(data[['district']])
data['district_num'] = pd.DataFrame(housing_category_encoded)
```

```python
def clean_data_2(df):
    df.drop('location',axis=1,inplace=True)
    df['price'] = df['price'].str.replace('Договорная', '71334').astype(int)
    df['size'] = df['size'].str.replace('Площадьземли:1сот', '100.0').astype(float)
    df['size_per_rooms'] = df['size'] / df['rooms']
    df['level_per_size'] = df['size'] / df['level']
    df['price_per_rooms'] = df['price'] / df['rooms']
    df['price_per_level'] = df['price'] / df['level']
    df['price_per_maxlevel'] = df['price'] / df['max_levels']
    df['price_per_size'] = df['price'] / df['size']
    df['max_levels_per_size'] = df['size'] / df['max_levels']
    df.drop('district', axis=1,inplace=True)
    return df
clean_data_2 = clean_data_2(data)
```

```python
clean_data_2.to_csv('clean_data.csv')
```

```python
clean_data_2.corrwith(clean_data_2['district_num']).sort_values(ascending=False)
```

```python
x = clean_data_2.drop("price",axis=1)
y = clean_data_2["price"]

X_train,x_test,Y_train,y_test = train_test_split(x,y, test_size=0.2)
```

```python
from sklearn.ensemble import BaggingRegressor

beggin = BaggingRegressor()
beggin.fit(X_train,Y_train)
```
