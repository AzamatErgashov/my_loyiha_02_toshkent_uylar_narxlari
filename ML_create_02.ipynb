```python
import pandas as pd
data = pd.read_csv('housing_data_08-02-2021.csv', index_col=0)
data.head()
```

```python
data['location'].value_counts()
```

```python
def clean_data(df):
    df.drop('location',axis=1,inplace=True)
    joylashuv_category_encoded = pd.get_dummies(df['district'])
    df.drop('district',axis=1,inplace=True)
    df['price'] = df['price'].str.replace('Договорная', '0')
    df['price'] = df['price'].fillna(method='bfill')
    df['price'] = df['price'].astype(int)
    df['size'] = df['size'].str.replace('Площадьземли:1сот', '0').astype(float)
    return np.c_[df, joylashuv_category_encoded]
data = clean_data(data)
```

```python
column = ['rooms','size','level','max_levels','price', 'Чиланзарский', 'Мирзо-Улугбекский', 'Юнусабадский', 'Учтепинский',
       'Мирабадский', 'Яккасарайский', 'Яшнободский', 'Шайхантахурский',
       'Олмазорский', 'Сергелийский', 'Янгихаётский', 'Бектемирский']
new_data = pd.DataFrame(data, columns=column)
```
